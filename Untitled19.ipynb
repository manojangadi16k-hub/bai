```python
from nltk.util import bigrams
import spacy
from nltk import ngrams
nlp = spacy.load("en_core_web_sm")
doc = nlp("A is simulation of human intelligence by machine.it include except system nlp speech regcognization and machine vision")
bigrams=ngrams(doc,2)
print(list(bigrams))
```

```python


import spacy
from nltk import trigrams
nlp= spacy.load("en_core_web_sm")
doc= nlp("A is simulation of human intelligence by machine.it include except system nlp speech regcognization and machine vision")
trigrams=trigrams(doc)
print(list(trigrams))
```
